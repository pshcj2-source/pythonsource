#gr.Blocks -- 레이아웃을 직접 설계하는 BMI 계산기

- gr.Interface의 "자동 배치"를 넘어서, gr.Blocks로 컴포넌트 배치를 직접 설계
- 버튼 클릭 이벤트(btn.click(...))로 원하는 타이밍에 함수를 실행시키는 법

```python
import gradio as gr

def calculate_bmi(height_cm, weight_g):
    height_m = height_cm / 100 # cm -> m단위 변환
    bmi = weight_g / (height_m **2)

# BMI 수치에 따라 결과 문구를 다르게 구성
    if bmi < 18.5:
        category = "저체중"
    elif bmi < 23:
        category = "정상"
    elif bmi < 25:
        category = "과체중"
    else:
        category = "비만"

    return f"{bmi:.1f}", category
```

```python
# gr.blocks() : 레이아웃을 직접 설계하는 컨테이너
with gr.Blocks(title="BMI 계산기") as demo:
    gr.Markdown("## BMI(체질량지수) 계산기") # 화면에 마크다운 텍스트 표시

    # Row: 컴퓨넌트 가로로 나란히 배치
    with gr.Row():
        height_input = gr.Number(label="키(cm)")
        weight_input = gr.Number(label="몸무게(kg)")

    calc_btn = gr.Button("계산하기") # 버튼 컨포넌트

# Row 안에 다시 두 개의 출력 컴포넌트를 나란히 배치
    with gr.Row():
        bmi_ouput = gr.Textbox(label="BMI 수치")
        category_output = gr.Textbox(label="체중 분류")

    # 버튼 클릭 이벤트
    # .click(함수, 입력 컴포넌트 리스트, 출력 컴퓨터넌트 리스트)
    # -> calc.btn을 클릭하면 calculate_bmi 함수를 height_input, weight_input 값으로 실행하고
    # 결과를 bmi_output, category_output에 넣어라

    calc_btn.click(
        fn=calculate_bmi,
        inputs=[height_input, weight_input],
        outputs=[bmi_ouput, category_output],
    )

demo.launch()
```
