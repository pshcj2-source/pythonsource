# 숫자 계산기 - 다양한 입력 컴퓨턴트 다루기

o gr.Number, gr.Radio 처럼 텍스트가 아닌 다양한 입력 컴포넌트를 이용

o 사용자가 화면에서 "선택"할 수 있는 옵션(라디오 버튼)을 만드는 법



```python
import gradio as gr

# 두 숫자와 연산자를 받아 계산 결과를 반환하는 함수
def calculate(num1, num2, operator):
    if operator == "덧셈 (+)":
        result = num1 + num2
    elif operator == "뺄셈 (-)":
        result = num1 - num2
    elif operator == "곱셈 (*)":
        result = num1 * num2
    elif operator == "나눗셈 (/)":
        # 0으로 나누는 경우를 미리 방어해줍니다.
        if num2 == 0:
            return "0으로는 나눌 수 없습니다!"
        result = num1 / num2
    else:
        result = "알 수 없는 연산자"
        
    return f"결과 : {result}"
```

```python
demo = gr.Interface(
    fn=calculate,
    inputs=[
        gr.Number(label="첫 번째 숫자"),
        gr.Number(label="두 번째 숫자"),
        # Radio : 여러 선택지 중 하나만 고르는 라디오 버튼
        gr.Radio(["덧셈 (+)", "뺄셈 (-)", "곱셈 (*)", "나눗셈 (/)"], label="연산자를 선택하세요"),
        #gr.Dropdown(["덧셈 (+)", "뺄셈 (-)", "곱셈 (*)", "나눗셈 (/)"], label="연산자를 선택하세요"),
    ],
    outputs=gr.Textbox(label="계산 결과"),
    title="간단한 숫자 계산기",
)
demo.launch()
```

```python
demo.close()
```
