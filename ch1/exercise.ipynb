```python
list1=["갑","을","병","정","경"]

#정 글자만 제외하고 출력 (for와 continue 사용)

for i in list1:
    if i != "정":
        print(i,end=" ")
print()
for i in list1:
    if i == "정": continue
    print(i,end=" ")
```

```python
scores=[70,60,55,75,95,90,80,85,100,78]

#10명 학생의 총점과 평균 출력

sum=0
for i in scores:
    sum+=i
print(sum)

print(sum/len(scores))
```
