```python
# 문자열을 입력받아 다음 정보를 출력
# 출력내용 : 입력한 문자, 각 문자의 유니코드 값, 가장 큰 문자, 가장 작은 문자

# 문장 입력 : Python  각 문자를 유니코드 값으로 표기   ord
# ex) P : 80, y : 121
# 가장 큰 문자 : y
# 가장 작은 문자 : P (대문자가 숫자 작음)
# ord,chr,max,min,list


word=input("영어 문자를 입력하세요")

# i=0
# while i < len(mz):
#     print(f'mz[i] : ord(mz[i])') 
#     i+=1

for ch in word:
    print(f'{ch} -> {ord(ch)}')

print(f'가장 큰 문자 : {max(word)}')

```

```python
x = [3,7,1,-5,9]
y=[2,-4,8,10,-1]
# zip, abs, enumerate  이용해서 거리 구하기

for i, (a,b) in enumerate(zip(x,y), start=1):
    distance = abs(a) + abs(b)
    print(f'{i}번 좌표 : ({a}, {b}) -> {distance}')

```

```python
# 사용자로부터 숫자 10개 입력받기

numbers = []
for i in range(10):
    num = int(input(f'숫자입력 : '))
    numbers.append(num)

# 출력
# 입력리스트, 중복제거, 오름차순, 튜플변환, 집합 크기


print(f"원본 리스트 {numbers}")
# 중복제거 {1,2,3,4} set()
unique_numbers = list(set(numbers))
print(f"중복 제거 {unique_numbers}")
# 정렬 sorted()
sorted_numbers=sorted(unique_numbers)
print(f"오름차순 정렬 {sorted_numbers}")
# 튜플 tuple()
print(f"리스트=>튜플 {tuple(sorted_numbers)}")
# 집합 크기 : len()
print(f'중복 제거 후 집합 크기 {len(unique_numbers)}')
```
