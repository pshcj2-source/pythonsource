#### 정규표현식(정규식)
- 복잡한 문자열을 처리할 때 사용하는 방법 (파이썬, 자바, 자바 스크립 등 대부분 프로그래밍 언어에서도 사용 가능)
- 파이썬만의 고유한 문법이 아니라 문자열을 처리하는 모든 곳에서 사용하는 일종의 형식 언어

| 매서드 | 목적 |
|-----| ----|
| match | 문자열의 처음부터 정규식과 매치되는지 조사 |
| search | 문자열 전체를 검색하여 정규식과 매치되는지 조사 |
| findall | 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴 |
| finditer | 정규식과 매치되는 모든 문자열(substring)을 이터레이터 객체로 리턴 |

```python
# 주민등록번호 -> 뒷자리를 모두 * 문자로 변경

# 파이썬에서 정규식 지원하는 모듈  #re
import re

dir(re)
```

```python

# re.compile() : 정규식을 작성

pattern = re.compile("ab*")   # re.compile : 패턴을 만든다.

# 패턴을 특정 문자열이 포함하고 있는가?
```

```python
# 패턴을 특정 문자열이 포함하고 있는가?
print(pattern.match("abc"))
```

```python
# [문자] : []안에 들어있는 각각 문자들 중 매치가 있는지 물어볼때

p = re.compile("[abc]")

print(p.match("a"))
print(p.match("before"))
print(p.match("dude"))
```

```python
p = re.compile("[abc]")  
m = p.match("cb")

print(f"패턴 시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"일치의 시작, 끝 반환 : {m.span()}")

if m:
    print(f"Match found {m.group()}")
else:
    print("No Match")

# re.compile("[abc]"):
# [abc]는 정규식에서 "a, b, c 중 문자 딱 1개"를 의미하는 패턴입니다.
# p.match("cb"):
# match() 메서드는 "문자열의 맨 처음 시작부터" 패턴과 일치하는지 확인합니다.
# 검사할 문자열은 "cb"입니다. 이 문자열의 첫 번째 글자는 'c'입니다.
# 패턴 [abc]에 'c'가 포함되어 있으므로, 맨 첫 글자에서 매칭이 성공합니다.
# 많은 분들이 match("cb")를 하면 문자열 전체인 "cb"가 다 나올 거라고 생각하기 쉽습니다.
# 하지만 [abc]는 글자 '하나'만 검사하는 패턴이기 때문입니다.
# 첫 번째 글자 'c'가 패턴에 맞으니 거기서 매칭이 성공하여 끝난 것이고, 뒤에 있는 'b'는 검사하기도 전에 이미 매치가 완료된 것입니다.
# 만약 "cb" 전체를 다 매치시키고 싶다면 패턴을 [abc]+ (1번 이상 반복) 형태로 작성해야 합니다.
```

```python
p = re.compile("[abc]")  
m = p.search("dc")  #search는 일치하는 것이 하나만 있어도 됨

print(m)
print(f"패턴 시작 : {m.start()}")
print(f"패턴 끝 : {m.end()}")
print(f"패턴과 일치하는 문자열 반환 : {m.group()}")
print(f"일치의 시작, 끝 반환 : {m.span()}")
```

```python
p = re.compile("[abcde]")  
print(p.search("3 before"))  # 문장 전체 중 맞는 것이 있어야 함
print(p.match("3 before"))   # 첫 시작부터 맞는것이 있어야함
                             # []는 안의 문자 중 일치하는 첫 글자 '하나'만 검사하는 패턴
```

```python
p = re.compile("[abcde]")  #a나 b나 c가 있는지 찾아줘
m= p.findall("Happy forever")   # 정규식과 매치되는 모든 문자열(substring)을 리스트로 리턴
print(m)

m= p.finditer("Happy forever")  # 정규식과 매치되는 모든 문자열(substring)을 이터레이터 객체로 리턴
print(m)
for r in m:
    print(r)

# re.Match object; span=(1,2), match='a'
# re.Match object; span=(9,10), match='e'
# re.Match object; span=(11,12), match='e'
```

```python
# "." 문자 : \n을 제외한 모든 문자

p = re.compile("a.b")  #'a'로 시작하고, 그 뒤에 아무 글자나 딱 1개('.')가 오고, 
                       # 마지막에 'b'로 끝나는 3글자 패턴을 의미
print(p.match("a\nb"))
print(p.match("abb"))
print(p.match("a0b"))
print(p.match("aab"))
print(p.match("a b"))
```

```python
#.:마침표로서 사용하고 싶을 떄
# \. or [.]
# \t, \r
p=re.compile(r"a\.b")
# p=re.compile(r"a[.]b")

print(p.match("a\nb"))
print(p.match("abb"))
print(p.match("a0b"))
print(p.match("aab"))
print(p.match("a b"))
```

```python
# * : 0 ~ 무한대
 
# ? : 0 ~ 1

# ca
p = re.compile("ca*t")
print(p.match("ct"))
print(p.match("cat"))
print(p.match("caaaaaaaaaaaaaat"))
```

```python
p = re.compile("ca?t")
print(p.match("ct"))
print(p.match("cat"))
print(p.match("caaaaaaaaaaaaaat"))
```

```python
p = re.compile("ca{2}t")
print(p.match("ct"))
print(p.match("ca1t"))
print(p.match("caaaaaaaaaaaaaat"))


p = re.compile("ca{2,5}t")
print(p.match("ct"))
print(p.match("cat"))
print(p.match("caat"))
print(p.match("caaaaaaaaaaaaaat"))
```

```python
# a~z 모든 소문자
p = re.compile("[a~z]")

print(p.search("kkk1234"))
print(p.search("k1234a12b"))
```

```python
# a~z 모든 소문자
p = re.compile("[a~z]+")

print(p.search("kkk1234"))
print(p.search("k1234a12b"))
```

```python
# [^ ] : 대괄호 안에 있는 문자를 제외한
# 영 대소문자, 숫자를 제외한 문자
p = re.compile("[^a-zA-Z0-9]+")

print(p.search("kkk1234"))
print(p.search("k1234a12b"))
print(p.search(" python"))

```

- 자주 사용하는 문자 표현식

| 정규표현식 | 설명 |
|-----| ----|
|\d|숫자와 매치 [0-9]와 동일한 표현식|
|\D|숫자가 아닌 것과 매치 [^0-9]와 동일한 표현식|
|\s|화이트스페이스(스페이스,탭,줄바꿈 등)문자와 매치 [빈칸\t\n\r\f\v]와 동일한 표현식|
|\S|화이트스페이스(스페이스,탭,줄바꿈 등)문자가 아닌 것과 매치 [^빈칸\t\n\r\f\v]와 동일한 표현식|
|\w|문자+숫자와 매치 [a-zA-Z0-9_]와 동일한 표현식|
|\W|문자+숫자가 아닌 것과 매치 [a-zA-Z0-9_]와 동일한 표현식|

```python
# re.I : ignorecase  대소문자 구별하지 않는거

# 영어가 포함되어 있는가?

p = re.compile("[a-z]+", re.I)
print(p.match("python"))
print(p.match("Python"))
print(p.match("PYTHON"))
```

```python
# re.M   : MULTILINE 인 것을 알려줌

data="""python one
life is too short
python two
you need python
python three
"""


# ^ : 대괄호 안에 드는건 "제외하는" 의미지만 대갈호 밖의 ^는 "시작하는" 의미가 있음
# python 시작하고 스페이스문자하나영어,숫자,_ 인 문자 찾아줘
p=re.compile(r"^python\s\w+",re.M)    # python으로 시작하고 
print(p.findall(data))
```

```python
# | : or

p=re.compile("Crow|Servo")
m=p.match("CrowHello")
print(m)

m=p.match("ServoHello")
print(m)
```

```python
#^ : 시작하는, $ 끝나는

p = re.compile("^Life")
m = p.search("Life is too short")
print(m)

m = p.search("My Life")
print(m)

p=re.compile("short$")
m = p.search("Life is too short")
print(m)
```

```python
# re.compile()
# re.sub()

# re.fullmatch() : 문자열 전체가 패턴과 일치

p = re.compile(r"\d+")
print(re.fullmatch(p, "1232abc"))
print(re.fullmatch(p, "1232456"))
print(re.fullmatch(p, "abc123"))
```

```python
# 영문 소문자로 시작하고 영문, 숫자 포함된 5~11글자

p=re.compile("[a-z][a-z0-9]{5~11}")
print(re.fullmatch(p, "1232abc"))
print(re.fullmatch(p, "1232456"))
print(re.fullmatch(p, "abc123"))

m=re.fullmatch(p, "abc123")
print(m.group())
```

```python
# 입력을 받은 값이 숫자인지 확인

# 숫자입니다. or 숫자가 아닙니다.

p = re.compile(r"\d+")
num = input("숫자를 입력하세요 : ")

if re.fullmatch(p, num):
    print("숫자입니다")
else:
    print("숫자가 아닙니다")
```
